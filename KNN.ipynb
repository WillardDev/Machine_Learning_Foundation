# K-Nearest Neighbors (KNN)

- This is a supervised machine learning algorithm generally used for classification but can also be used for regression tasks.
- It is a lazy learning algorithm, meaning it generalizes data only after it has been trained.
- KNN is a distance-based algorithm, meaning it uses the distance between data points to make predictions.
- KNN is a non-parametric algorithm, meaning it does not make any assumptions about the underlying data distribution.
- KNN is a simple and easy-to-implement algorithm.
