# Support Vector Machines

```python
# import libraries
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# load dataset
data = load_breast_cancer()
X = data.data
y = data.target

feature_names = data.feature_names
print('Feature Shape:',X.shape)
print('Target Shape:',y.shape)
```

```python
print('Feature Names:',feature_names)
```

```python
# select two features for visualization (mean radius and mean texture)
X_Vis = X[:, [0, 1]]
feature_X, feature_y = feature_names[0], feature_names[1]
```

```python
# train test split
X_train, X_test, y_train, y_test = train_test_split(X_Vis, y, test_size=0.8, random_state=42)
```

```python
# scale the features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# train the model
svc = SVC(kernel='linear', C=1.0, random_state=42)
svc.fit(X_train, y_train)
```

```python
# evaluate the model
print('Training Accuracy:', svc.score(X_train, y_train))
print("Testing Accuracy:", svc.score(X_test, y_test))
```

```python
# plot decision boundary
plt.figure(figsize=(8, 6))
DecisionBoundaryDisplay.from_estimator(
    svc, X_train, response_method="predict", xlabel=feature_X, ylabel=feature_y
)
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap=plt.cm.inferno, s=20, edgecolor="k") 
```
